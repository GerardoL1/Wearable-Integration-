# Morning readiness: Ridge regression

Predict the self-reported morning answer to **“How ready do you feel to train hard today?”**, on a 1–10 scale.
One example is one person and the sleep session ending on that morning's local date.

**Current status: implementation ready; no model fitted.** The supplied CSVs have no participant IDs, timezone,
HRV, or readiness labels. They cannot be turned into labeled examples by guessing identity or inventing a score.
This notebook audits those files and implements a reusable pipeline for a verified nightly table.
It never edits either original CSV and never uses a vendor recovery score as the target.

Run cells from top to bottom. The audit runs now. Training and prediction activate only when the documented
nightly input is available and the training switch is enabled. No synthetic labels are added to real data.

````python
DEFAULT_CONFIG = {
    "raw_features": ["hrv", "rhr", "total_sleep_min"],
    "derived_features": ["z_hrv", "rhr_delta", "sleep_debt_min", "hrv_trend"],
    "lookback_days": 28,
    "min_history": 7,
    "approved_hrv_method": "SDNN",
    "allow_mixed_devices": False,
    "approved_devices": [],
    "alpha_grid": [0.1, 1.0, 10.0, 100.0],
    "outer_splits": 5,
    "inner_splits": 4,
}

RAW_FEATURES = {
    "hrv", "rhr", "total_sleep_min", "time_in_bed_min", "efficiency", "awakenings",
    "deep_min", "core_min", "rem_min", "training_load", "respiratory_rate", "skin_temp",
}
DEPENDENCIES = {"z_hrv": "hrv", "rhr_delta": "rhr",
                "sleep_debt_min": "total_sleep_min", "hrv_trend": "hrv"}
CORE_FEATURES = {"hrv", "rhr", "total_sleep_min"}


def check_config(config):
    raw, derived = config["raw_features"], config["derived_features"]
    if not raw or len(set(raw)) != len(raw) or not set(raw) <= RAW_FEATURES:
        raise ValueError("raw_features must be unique supported sensor features.")
    if not set(raw) & CORE_FEATURES:
        raise ValueError("Select at least one core feature: hrv, rhr, total_sleep_min.")
    if len(set(derived)) != len(derived) or not set(derived) <= set(DEPENDENCIES):
        raise ValueError("Unsupported or repeated derived feature.")
    for feature in derived:
        if DEPENDENCIES[feature] not in raw:
            raise ValueError(f"{feature} requires {DEPENDENCIES[feature]} in raw_features.")
    if config["lookback_days"] != 28 or config["min_history"] != 7:
        raise ValueError("This implementation fixes the agreed 28-day / 7-observation baseline.")
    if config["approved_hrv_method"] not in {"SDNN", "RMSSD"}:
        raise ValueError("Choose one HRV method explicitly; SDNN and RMSSD cannot be mixed.")
    if config["outer_splits"] < 3 or config["inner_splits"] < 2:
        raise ValueError("Use >=3 outer folds and >=2 inner folds for nested grouped validation.")
    if not config["alpha_grid"] or any(not np.isfinite(a) or a <= 0 for a in config["alpha_grid"]):
        raise ValueError("alpha_grid must contain positive finite values.")
````

````python
CONFIG = copy.deepcopy(DEFAULT_CONFIG)
NIGHTLY_PATH = PROJECT_ROOT / "training_table.csv"
OUTPUT_DIR = PROJECT_ROOT / "ridge_outputs"
RUN_TRAINING = False

# Review these choices before enabling training. Do not silently remove HRV or mix devices.
# A reduced-input model requires an explicit raw_features / derived_features change.
# To approve multiple known devices, set allow_mixed_devices=True AND enumerate approved_devices.
check_config(CONFIG)
````

## 1. Inspect the original files

This audit describes records rather than assuming that device names identify people. Units for Heart Rate
are inferred as bpm; the raw files contain no explicit unit metadata. Datetimes have no timezone offset.
243 distinct sleep end dates do not establish 243 person-nights. Licence and provenance must be obtained
from the actual Kaggle page before reuse. Apple HealthKit's SDNN field is documented at
https://developer.apple.com/documentation/healthkit/hkquantitytypeidentifier/heartratevariabilitysdnn .
There is no HRV column in these CSVs; sampled bpm is not a substitute for beat-to-beat variability.

````python
def audit_original_files(root):
    summaries, frames = [], {}
    for name in ("heart_rate_data.csv", "sleep_data.csv"):
        path = Path(root) / name
        frame = pd.read_csv(path)
        frames[name] = frame
        times = [c for c in frame if c in {"Timestamp", "Start Time", "End Time"}]
        for column in times:
            parsed = pd.to_datetime(frame[column], errors="raise")
            summaries.append({"file": name, "rows": len(frame), "column": column,
                              "first": str(parsed.min()), "last": str(parsed.max())})
        print(f"{name}: {len(frame):,} rows; columns: {', '.join(frame.columns)}")
        print("  Devices:", sorted(frame["Source Name"].dropna().unique().tolist()))
        print("  Blank counts:", frame.isna().sum().to_dict())
    sleep = frames["sleep_data.csv"]
    start, end, stamp = [pd.to_datetime(sleep[c]) for c in ("Start Time", "End Time", "Timestamp")]
    overlaps = 0
    for _, part in sleep.assign(_start=start, _end=end).groupby("Source Name"):
        ordered = part.sort_values("_start")
        overlaps += int((ordered["_start"] < ordered["_end"].cummax().shift()).sum())
    print("Sleep categories:", sleep["Category"].value_counts().to_dict())
    print("Distinct recorded end dates:", end.dt.date.nunique())
    print("Sleep intervals overlapping earlier intervals within device:", overlaps)
    print("Associated HR timestamps outside interval:", int(((stamp < start) | (stamp > end)).sum()))
    print("NOT TRAINED: participant IDs, timezone, HRV, morning labels, and licence are absent.")
    return pd.DataFrame(summaries)
````

## 2. Supply a verified nightly table

Create `training_table.csv` only after the missing identity, local-time, label and measurement questions are
resolved. Its header for the default model is:

```text
tester_id,night_date,device_name,hrv_method,hrv,rhr,total_sleep_min,ready_rating
```

| Column | Contract |
|---|---|
| tester_id | Pseudonym such as P01; one actual participant, not one device model |
| night_date | YYYY-MM-DD; local morning on which the sleep session ends |
| device_name | Verified source device; explicit approval is needed before pooling models |
| hrv_method | SDNN by default; one method only, never silently converted from RMSSD |
| hrv | Overnight HRV in ms; freeze and document its aggregation method before collection |
| rhr | bpm; freeze either exported resting HR or lowest observed overnight HR, and keep the definition consistent |
| total_sleep_min | Minutes asleep after resolving overlaps and unspecified intervals |
| ready_rating | Real morning integer 1–10; blank on unlabeled history rows |

Do not add unsupported columns full of zero. Optional measurements can be explicitly selected in CONFIG
only after they are actually collected and defined. `training_load` must already refer to the previous
local day's exercise; no current-day workout information may enter a morning prediction.
If a survey table is joined separately, join its local morning date to night_date, not to the prior evening's date.
All input measurements must have been available by the prediction cutoff. No guessed timezone conversion
or raw sleep aggregation is performed here because the source metadata is missing.

Input IDs and dates are keys, never predictors. Existing derived features and has_* columns are recomputed.

````python
def validate_nightly(data, config, require_target=False):
    check_config(config)
    frame = data.copy(deep=True)
    required = {"tester_id", "night_date", "device_name", *config["raw_features"]}
    if require_target:
        required.add("ready_rating")
    missing = sorted(required - set(frame.columns))
    if missing:
        raise ValueError("Missing nightly columns: " + ", ".join(missing) +
                         ". The raw CSVs are not a verified labeled person-night table.")
    if frame.empty:
        raise ValueError("The nightly table is empty.")
    for key in ("tester_id", "device_name"):
        if frame[key].isna().any() or frame[key].astype(str).str.strip().eq("").any():
            raise ValueError(f"{key} cannot be blank.")
        frame[key] = frame[key].astype(str).str.strip()
    dates = frame["night_date"].astype(str)
    if not dates.str.fullmatch(r"\d{4}-\d{2}-\d{2}").all():
        raise ValueError("night_date must be a verified local YYYY-MM-DD morning date.")
    frame["night_date"] = pd.to_datetime(dates, format="%Y-%m-%d", errors="raise")
    if frame.duplicated(["tester_id", "night_date"]).any():
        raise ValueError("Duplicate tester_id / night_date rows; resolve sessions before fitting.")
    devices = set(frame["device_name"])
    approved = set(config["approved_devices"])
    if approved and not devices <= approved:
        raise ValueError("An unapproved device is present.")
    if len(devices) > 1 and (not config["allow_mixed_devices"] or not devices <= approved):
        raise ValueError("Multiple devices need explicit pooling approval and approved_devices entries.")
    if "ready_rating" not in frame:
        frame["ready_rating"] = np.nan
    target = pd.to_numeric(frame["ready_rating"], errors="raise")
    observed = target.notna()
    if ((~np.isfinite(target[observed])) | (~target[observed].between(1, 10)) |
            target[observed].mod(1).ne(0)).any():
        raise ValueError("ready_rating must be a real integer morning answer from 1 to 10.")
    if require_target and not observed.any():
        raise ValueError("No ready_rating labels exist; training is not possible.")
    frame["ready_rating"] = target
    flagged = {}
    for feature in config["raw_features"]:
        values = pd.to_numeric(frame[feature], errors="raise").astype(float)
        bad = values.notna() & ~np.isfinite(values)
        if feature == "hrv":
            bad |= values.notna() & ~values.between(5, 250)
        elif feature == "rhr":
            bad |= values.notna() & ~values.between(30, 120)
        elif feature == "total_sleep_min":
            bad |= values.notna() & ~values.between(120, 840)
        elif feature == "efficiency":
            bad |= values.notna() & ~values.between(0, 1)
        elif feature in {"time_in_bed_min", "deep_min", "core_min", "rem_min", "training_load"}:
            bad |= values.notna() & (values < 0)
        elif feature == "awakenings":
            bad |= values.notna() & ((values < 0) | values.mod(1).ne(0))
        frame[f"invalid_{feature}"] = bad.astype(int)
        flagged[feature] = int(bad.sum())
        frame[feature] = values.mask(bad)
    if "hrv" in config["raw_features"]:
        if "hrv_method" not in frame:
            raise ValueError("hrv_method is required; SDNN and RMSSD must not be mixed.")
        methods = frame.loc[frame["hrv"].notna(), "hrv_method"].fillna("").astype(str).str.upper().str.strip()
        if not methods.eq(config["approved_hrv_method"]).all():
            raise ValueError("HRV method is missing or differs from the approved method; no SDNN/RMSSD mixing.")
    frame = frame.sort_values(["tester_id", "night_date"]).reset_index(drop=True)
    report = {"people": int(frame.tester_id.nunique()), "nights": len(frame),
              "nights_with_target": int(frame.ready_rating.notna().sum()),
              "rows_removed": 0, "values_flagged_and_emptied": flagged,
              "missing_after_flags": frame[config["raw_features"] + ["ready_rating"]].isna().sum().to_dict()}
    return frame, report


def build_features(frame, config):
    check_config(config)
    groups = []
    core = sorted(set(config["raw_features"]) & CORE_FEATURES)
    for _, part in frame.groupby("tester_id", sort=False):
        group = part.sort_values("night_date").copy().set_index("night_date", drop=False)
        means, counts = {}, {}
        for feature in core:
            series = group[feature]
            window = series.rolling("28D", closed="left", min_periods=7)
            means[feature] = window.mean()
            counts[feature] = series.rolling("28D", closed="left", min_periods=0).count()
        if "z_hrv" in config["derived_features"]:
            std = group.hrv.rolling("28D", closed="left", min_periods=7).std(ddof=0)
            group["z_hrv"] = (group.hrv - means["hrv"]) / std.where(std > 1e-12)
        if "rhr_delta" in config["derived_features"]:
            group["rhr_delta"] = group.rhr - means["rhr"]
        if "hrv_trend" in config["derived_features"]:
            recent = group.hrv.rolling("7D", closed="left", min_periods=7).mean()
            group["hrv_trend"] = recent / means["hrv"].where(means["hrv"] > 0)
        if "sleep_debt_min" in config["derived_features"]:
            deficits = means["total_sleep_min"] - group.total_sleep_min
            group["sleep_debt_min"] = deficits.rolling("3D", closed="left", min_periods=3).sum()
        elapsed = (group.index - group.index.min()).days >= 7
        history_ok = np.logical_and.reduce([counts[f].ge(7).to_numpy() for f in core])
        group["baseline_ready"] = (elapsed & history_ok).astype(int)
        # Blank current-night values can be imputed, but a night with no selected core signal gets no score.
        group["current_data_ready"] = group[core].notna().any(axis=1).astype(int)
        for feature in config["raw_features"] + config["derived_features"]:
            group[f"has_{feature}"] = group[feature].notna().astype(int)
        groups.append(group.reset_index(drop=True))
    return pd.concat(groups, ignore_index=True)
````

## 3. Feature definitions and exclusions

The proposed, reviewable baseline convention is the **28 calendar nights before the current night**,
with at least 7 valid observations per selected core signal. Missing calendar nights consume time in the
lookback; they are not replaced with old observations. The first 7 calendar days are baseline_ready=0.
Current-night measurements can be predictors after waking, but never enter their own historical reference.

* z_hrv = (current HRV − prior mean) / prior population SD; zero SD leaves this feature empty.
* rhr_delta = current overnight/resting HR − prior mean HR.
* hrv_trend = mean of the 7 previous calendar nights / mean of the 28 previous calendar nights.
  Seven valid observations are required in the short window; the current night is excluded.
* sleep_debt_min = sum of the **three previous calendar nights' signed sleep deficits**.
  Each earlier deficit is its own prior-28-night mean sleep minus its sleep duration. Surplus offsets deficit.
  The current night is excluded, matching the strict past-only requirement. Missing constituent nights leave
  this value empty. It becomes available later than the initial baseline warmup.

These definitions are documented choices for review, not a claim that surplus sleep fully reverses recovery debt.
If the team wants current-night debt or zero-clipped deficits, approve that definition before changing it.
Invalid sensor values are flagged and set to NaN; no participant is dropped. Labels are never imputed.
Rows without a target, sufficient history, or any current core signal are retained in the feature table but
withheld from fitting. No target, ID, date, device, or free-text column enters the predictor matrix.

````python
nightly_data = None
feature_table = None
quality_report = None
if NIGHTLY_PATH.exists():
    nightly_data = pd.read_csv(NIGHTLY_PATH, dtype={"tester_id": "string"})
    validated, quality_report = validate_nightly(nightly_data, CONFIG)
    feature_table = build_features(validated, CONFIG)
    display(pd.DataFrame([quality_report]))
    display(feature_table.head(10))
    display(feature_table[CONFIG["raw_features"] + CONFIG["derived_features"]].describe().T)
    print("Baseline-ready nights:", int(feature_table.baseline_ready.sum()))
else:
    print("NOT TRAINED: training_table.csv does not exist. Supply verified person-night data and real ratings.")
````

## 4. Fit and evaluate Ridge

The estimator is a single pipeline: training-fold median imputation → StandardScaler → Ridge.
Explicit has_* flags accompany every selected raw and derived feature. An entirely unobserved selected
feature in a training fold stops fitting instead of inventing a zero or silently changing the model.

Nested GroupKFold keeps participants disjoint: inner folds choose alpha; outer folds estimate performance
on unseen people. At least 3 **eligible labeled** people are required. Each fold's median DummyRegressor
is a comparison baseline. Tuning minimizes equal-weight per-person MAE, with predictions bounded to 1–10
consistently during tuning, evaluation and inference. Raw predictions are retained for diagnostics.
There is no random row split or early-stopping split. All earlier sensor history, including unlabeled history,
can construct features for held-out participants: this evaluates users after sensor-only onboarding.

Reported nested CV is not a separately frozen external test. For a release decision, reserve additional people
before tuning and evaluate that final cohort once. A one-person dataset cannot support person-held-out validation.

````python
def predictor_columns(config):
    sensors = config["raw_features"] + config["derived_features"]
    return sensors + [f"has_{feature}" for feature in sensors]


def make_estimator(alpha):
    return Pipeline([
        ("imputer", SimpleImputer(strategy="median", keep_empty_features=False)),
        ("scaler", StandardScaler()),
        ("ridge", Ridge(alpha=float(alpha))),
    ])


def checked_fit(model, X, y):
    empty = X.columns[X.isna().all()].tolist()
    if empty:
        raise ValueError("Selected features have no observations in a training fold: " +
                         ", ".join(empty) + ". Collect data or explicitly approve another feature set.")
    if np.isinf(X.to_numpy(dtype=float)).any():
        raise ValueError("Infinite predictor values are not allowed.")
    return model.fit(X, y)


def bounded_predict(model, X):
    return np.clip(model.predict(X), 1.0, 10.0)


def person_mae(y, prediction, groups):
    values = pd.DataFrame({"person": np.asarray(groups),
                           "error": np.abs(np.asarray(y) - np.asarray(prediction))})
    return float(values.groupby("person").error.mean().mean())


def tune_alpha(X, y, groups, config):
    count = pd.Series(groups).nunique()
    if count < 2:
        raise ValueError("At least two training participants are required for inner grouped tuning.")
    splitter = GroupKFold(n_splits=min(config["inner_splits"], count))
    scores = []
    for alpha in config["alpha_grid"]:
        predictions = np.full(len(y), np.nan)
        for train_idx, valid_idx in splitter.split(X, y, groups):
            model = checked_fit(make_estimator(alpha), X.iloc[train_idx], y.iloc[train_idx])
            predictions[valid_idx] = bounded_predict(model, X.iloc[valid_idx])
        scores.append({"alpha": float(alpha), "person_mae": person_mae(y, predictions, groups)})
    best = min(scores, key=lambda item: (item["person_mae"], -item["alpha"]))
    return best["alpha"], scores


def train_ridge(table, config):
    check_config(config)
    if "ready_rating" not in table or not table.ready_rating.notna().any():
        raise ValueError("No ready_rating labels exist; no model was fitted.")
    eligible = table.baseline_ready.eq(1) & table.current_data_ready.eq(1) & table.ready_rating.notna()
    data = table.loc[eligible].reset_index(drop=True)
    all_people = set(table.tester_id)
    usable_people = set(data.tester_id)
    if all_people - usable_people:
        raise ValueError("Some participants have no eligible labeled nights: " +
                         ", ".join(sorted(all_people - usable_people)) +
                         ". Collect labels/history or explicitly approve their exclusion; no person was dropped.")
    if len(usable_people) < 3:
        raise ValueError("Need at least 3 eligible labeled participants for nested GroupKFold; no random-row fallback.")
    columns = predictor_columns(config)
    X, y, groups = data[columns], data.ready_rating.astype(float), data.tester_id
    outer = GroupKFold(n_splits=min(config["outer_splits"], len(usable_people)))
    raw_pred, pred, dummy_pred = [np.full(len(data), np.nan) for _ in range(3)]
    folds = np.full(len(data), -1, dtype=int)
    fold_audit = []
    for fold, (train_idx, test_idx) in enumerate(outer.split(X, y, groups), start=1):
        train_people = set(groups.iloc[train_idx])
        test_people = set(groups.iloc[test_idx])
        assert train_people.isdisjoint(test_people)
        alpha, scores = tune_alpha(X.iloc[train_idx].reset_index(drop=True),
                                  y.iloc[train_idx].reset_index(drop=True),
                                  groups.iloc[train_idx].reset_index(drop=True), config)
        model = checked_fit(make_estimator(alpha), X.iloc[train_idx], y.iloc[train_idx])
        raw_pred[test_idx] = model.predict(X.iloc[test_idx])
        pred[test_idx] = np.clip(raw_pred[test_idx], 1, 10)
        dummy = DummyRegressor(strategy="median").fit(X.iloc[train_idx], y.iloc[train_idx])
        dummy_pred[test_idx] = bounded_predict(dummy, X.iloc[test_idx])
        folds[test_idx] = fold
        fold_audit.append({"fold": fold, "alpha": alpha,
                           "train_people": sorted(train_people), "test_people": sorted(test_people),
                           "inner_scores": scores})
    oof = data[["tester_id", "night_date", "device_name", "ready_rating"]].copy()
    oof["fold"], oof["raw_prediction"] = folds, raw_pred
    oof["predicted_ready_rating"], oof["median_baseline_prediction"] = pred, dummy_pred
    oof["absolute_error"] = np.abs(y - pred)
    oof["baseline_absolute_error"] = np.abs(y - dummy_pred)
    per_person = oof.groupby("tester_id").agg(
        nights=("ready_rating", "size"), mae=("absolute_error", "mean"),
        baseline_mae=("baseline_absolute_error", "mean"))
    final_alpha, final_scores = tune_alpha(X, y, groups, config)
    final_model = checked_fit(make_estimator(final_alpha), X, y)
    feature_sensors = config["raw_features"] + config["derived_features"]
    correlations = {}
    for feature in feature_sensors:
        pair = data[[feature, "ready_rating"]].dropna()
        correlations[feature] = (float(pair[feature].corr(pair.ready_rating))
                                 if len(pair) >= 3 and pair[feature].nunique() > 1 and
                                 pair.ready_rating.nunique() > 1 else None)
    metrics = {
        "evaluation": "nested person-group cross-validation; not an external final test",
        "people": len(usable_people), "eligible_labeled_nights": len(data),
        "retained_input_nights": len(table), "withheld_from_fitting": int((~eligible).sum()),
        "withheld_missing_target": int(table.ready_rating.isna().sum()),
        "withheld_baseline_unready": int(table.baseline_ready.eq(0).sum()),
        "withheld_no_current_core_signal": int(table.current_data_ready.eq(0).sum()),
        "exclusion_counts_overlap": True,
        "mae_points": float(np.abs(y - pred).mean()),
        "person_mae_points": person_mae(y, pred, groups),
        "rmse_points": float(np.sqrt(np.mean((y - pred) ** 2))),
        "within_one_point": float(np.mean(np.abs(y - pred) <= 1)),
        "median_baseline_person_mae": person_mae(y, dummy_pred, groups),
        "per_person": per_person.reset_index().to_dict("records"),
        "per_device_mae": oof.groupby("device_name").absolute_error.mean().to_dict(),
        "target_histogram": {str(i): int(y.eq(i).sum()) for i in range(1, 11)},
        "feature_target_correlations_quick_check": correlations,
        "outer_fold_audit": fold_audit, "final_alpha": final_alpha,
        "final_inner_scores": final_scores,
    }
    return {"model": final_model, "config": copy.deepcopy(config), "feature_columns": columns,
            "approved_devices": sorted(set(table.device_name)), "metrics": metrics,
            "oof_predictions": oof,
            "versions": {name: importlib.metadata.version(name)
                         for name in ("numpy", "pandas", "scikit-learn", "joblib")}}
````

````python
bundle = None
if not RUN_TRAINING:
    print("NOT TRAINED: RUN_TRAINING=False. Review data/configuration, then enable it when real labels exist.")
elif feature_table is None:
    print("NOT TRAINED: verified training_table.csv is missing. Original CSVs contain no readiness target.")
else:
    # Raises an actionable error instead of fitting if labels, groups, method, or coverage are unsuitable.
    validated, quality_report = validate_nightly(nightly_data, CONFIG, require_target=True)
    feature_table = build_features(validated, CONFIG)
    bundle = train_ridge(feature_table, CONFIG)
    display(pd.DataFrame([{k: v for k, v in bundle["metrics"].items()
                           if isinstance(v, (str, int, float, bool))}]))
    display(bundle["oof_predictions"].head(10))
    display(pd.DataFrame(bundle["metrics"]["per_person"]))
    print("Per-person MAE must beat the median baseline before considering release.")
````

## 5. Save the fitted pipeline and predict

This section writes artifacts only after a successful real-data fit. The saved bundle includes preprocessing,
feature selection, alpha, baseline conventions, accepted devices/method, and package versions. Predictions
require the same verified nightly schema and earlier sensor history; missing targets are allowed at inference.
Withheld nights have blank predictions and an explicit status, never a made-up score.

For inference later: load your own trusted `ridge_outputs/ridge_readiness.joblib`, read a nightly table with
at least the preceding **31 calendar nights** of available sensor history, and call
`predict_readiness(loaded_bundle, history_and_current_nights)`. Select the latest morning from the returned table.
The extra three days let the previous three sleep deficits each use their own complete 28-night baseline.
Provide all available history during initial onboarding. Truncating a mature user's history to only 28 nights
would change sleep debt and can change predictions. History and current records must have unique person-night keys.
The full selected feature schema is required;
individual missing measurements may be empty. Unseen devices require approval and revalidation.

````python
def predict_readiness(bundle, nightly_frame):
    config = copy.deepcopy(bundle["config"])
    config["approved_devices"] = bundle["approved_devices"]
    validated, _ = validate_nightly(nightly_frame, config, require_target=False)
    table = build_features(validated, config)
    output = table[["tester_id", "night_date", "baseline_ready", "current_data_ready"]].copy()
    output["raw_prediction"] = np.nan
    output["predicted_ready_rating"] = np.nan
    output["prediction_status"] = "collecting_baseline"
    output.loc[table.current_data_ready.eq(0), "prediction_status"] = "missing_current_core_signals"
    eligible = table.baseline_ready.eq(1) & table.current_data_ready.eq(1)
    if eligible.any():
        X = table.loc[eligible, bundle["feature_columns"]]
        raw = bundle["model"].predict(X)
        output.loc[eligible, "raw_prediction"] = raw
        output.loc[eligible, "predicted_ready_rating"] = np.clip(raw, 1, 10)
        output.loc[eligible, "prediction_status"] = "scored"
    return output


def save_artifacts(bundle, feature_table, quality_report, output_dir):
    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)
    joblib.dump(bundle, output_dir / "ridge_readiness.joblib")
    feature_table.to_csv(output_dir / "training_features.csv", index=False)
    bundle["oof_predictions"].to_csv(output_dir / "out_of_fold_predictions.csv", index=False)
    (output_dir / "metrics.json").write_text(json.dumps(bundle["metrics"], indent=2, allow_nan=False), encoding="utf-8")
    (output_dir / "quality_report.json").write_text(json.dumps(quality_report, indent=2), encoding="utf-8")
    (output_dir / "model_config.json").write_text(json.dumps({
        "config": bundle["config"], "feature_columns": bundle["feature_columns"],
        "approved_devices": bundle["approved_devices"], "versions": bundle["versions"]}, indent=2), encoding="utf-8")
````

````python
if bundle is not None:
    save_artifacts(bundle, feature_table, quality_report, OUTPUT_DIR)
    print("Saved fitted pipeline and evaluation artifacts to", OUTPUT_DIR)
    print("Latest-night predictions below are inference examples, not an additional test result:")
    predictions = predict_readiness(bundle, nightly_data)
    display(predictions.sort_values("night_date").groupby("tester_id", as_index=False).tail(1))
else:
    print("No fitted model or prediction artifacts written: a valid labeled cohort is required.")

for name, expected_hash in ORIGINAL_HASHES.items():
    assert hashlib.sha256((PROJECT_ROOT / name).read_bytes()).hexdigest() == expected_hash
print("Original CSVs are unchanged.")
````

## Verification and limits

Run `python -m unittest discover -s tests -v` in an environment with the dependencies installed.
The tests exercise leakage, warmup, missingness, identity/device guards, grouped tuning, serialization,
and prediction using isolated synthetic fixtures. Those fixtures never label or train on the original CSVs.

This notebook is a proposed Ridge implementation, not a validated readiness score. A trained artifact and
honest real-world performance cannot exist until linked morning ratings and participant metadata are supplied.
Feature correlations are descriptive only; repeated nights are dependent. Review errors by person and by
rating range. Final testing should use reserved people; chronological testing answers a separate question
about future nights for existing users. Changes to target, participant exclusion, device pooling, reduced
feature sets, or baseline definitions need the team's explicit decision before fitting.
